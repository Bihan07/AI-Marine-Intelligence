```python
import pandas as pd

df = pd.read_csv("../data/raw/obis/occurrences.csv")

df.head()
```

```python
df.columns.tolist()
```

```python
df.isnull().sum().sort_values(ascending=False).head(15)
```

```python
useful_columns = [
    "scientificName",
    "species",
    "genus",
    "family",
    "decimalLatitude",
    "decimalLongitude",
    "eventDate",
    "depth",
    "sst",
    "sss",
    "individualCount"
]

df_clean = df[
    [col for col in useful_columns if col in df.columns]
]

df_clean.head()
```

```python
from pathlib import Path

output_dir = Path("../data/processed/obis")
output_dir.mkdir(parents=True, exist_ok=True)

df_clean.to_csv(
    output_dir / "obis_clean.csv",
    index=False
)

print("Processed dataset saved!")
```
